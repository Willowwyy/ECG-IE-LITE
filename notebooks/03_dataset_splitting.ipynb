```python
import json
import random
import os
```

```python
# 配置路径
SOURCE_FILE = '../data/processed/train_dataset_raw.jsonl'
TRAIN_FILE = '../data/processed/train.jsonl'
TEST_FILE = '../data/processed/test.jsonl'
```

```python
# 1. 读取所有数据
with open(SOURCE_FILE, 'r', encoding='utf-8') as f:
    lines = f.readlines()

# 去除可能的空行
lines = [line.strip() for line in lines if line.strip()]

print(f"原始数据总共: {len(lines)} 条")
```

```python
# 2. 随机打乱
random.seed(42) # 保证每次打乱顺序一样，可复现
random.shuffle(lines)
```

```python
# 3. 切分 (90% 训练, 10% 测试)
split_index = int(len(lines) * 0.9)
train_data = lines[:split_index]
test_data = lines[split_index:]
```

```python
# 4. 保存
with open(TRAIN_FILE, 'w', encoding='utf-8') as f:
    f.write('\n'.join(train_data))

with open(TEST_FILE, 'w', encoding='utf-8') as f:
    f.write('\n'.join(test_data))

print(f"✅ 切分完成！")
print(f" - 训练集 (给模型学): {len(train_data)} 条 -> {TRAIN_FILE}")
print(f" - 测试集 (留着考试): {len(test_data)} 条 -> {TEST_FILE}")
```
